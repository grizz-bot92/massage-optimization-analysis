```python
import pandas as pd
```

```python
df_2021 = pd.read_csv("../data/raw/massage2021.csv")
df_2022 = pd.read_csv("../data/raw/massage2022.csv")
df_2023 = pd.read_csv("../data/raw/massage2023.csv")
df_2024 = pd.read_csv("../data/raw/massage2024.csv")
```

Data Sources: Data is split between four CSV exports from 2021-2024. 2024 only includes partial data.

```python
df_all = pd.concat(
    [df_2021, df_2022, df_2023, df_2024],
    ignore_index=True
)
```

```python
df_all.shape
df_all['Status'].value_counts()
df_all['Appointment On'].min(), df_all['Appointment On'].max()
```

```python
df_all.isna().sum()
```

```python
df_all["appointment_dt"] = pd.to_datetime(
    df_all['Appointment On'],
    errors="coerce"
)

df_all["created_dt"] = pd.to_datetime(
    df_all["Date Created"],
    errors="coerce"
)
```
